```python
print("Name: Parashuram Vaddekunta")
print("Project: Cars24 Data Analysis\n")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries loaded successfully!")

dataset_url = "https://raw.githubusercontent.com/jatin5jan/CarPredict/main/cars_24_combined.csv"
df = pd.read_csv(dataset_url)
print("Dataset loaded successfully!\n")

print("First 5 rows of the dataset:")
display(df.head())
```

## 1. Basic Dataset Details

```python
print("--- Dataset Size ---")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\n--- Column Names ---")
print(df.columns.tolist())

print("\n--- Column Types and Non-Empty Counts ---")
df.info()

numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(include=['object', 'category']).columns.tolist()

print("\nNumber Columns:", numerical_columns)
print("Text Columns:", categorical_columns)
```

## 2. Check Missing Values & Duplicate Rows

```python
print("--- Missing Values Check ---")
missing_count = df.isnull().sum()
missing_percent = (missing_count / len(df)) * 100

missing_table = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing %': missing_percent
}).sort_values(by='Missing Count', ascending=False)

display(missing_table)

print("\n--- Duplicate Rows Check ---")
duplicate_count = df.duplicated().sum()
print("Total Duplicate Rows:", duplicate_count)

if duplicate_count > 0:
    display(df[df.duplicated()])
else:
    print("No duplicate rows found.")
```

## 3. Summary Numbers & Category Counts

```python
print("--- Overall Data Summary ---")
display(df.describe(include='all'))

print("\n--- Count of Values in Each Text Column ---")
for col in categorical_columns:
    print(f"\nCounts for {col}:")
    print(df[col].value_counts())
```

## 4. Single Column Charts (Univariate Analysis)

```python
df.select_dtypes(include=['int64', 'float64']).hist(
    figsize=(15, 12),
    bins=30,
    edgecolor='black'
)
plt.suptitle('Distribution of All Number Columns', y=1.02)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
sns.histplot(df['Price'], kde=True, bins=30)
plt.title('Distribution of Car Prices')
plt.xlabel('Price')
plt.ylabel('Number of Cars')
plt.tight_layout()
plt.show()

for col in categorical_columns:
    plt.figure(figsize=(10, 4))
    sns.countplot(data=df, x=col, order=df[col].value_counts().index)
    plt.title(f'Count of Cars by {col}')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

## 5. Outlier Detection (Boxplots & IQR Method)

```python
for col in numerical_columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[col])
    plt.title(f'Outlier Check for {col}')
    plt.tight_layout()
    plt.show()

print("\n--- Outlier Count in Each Number Column ---")
for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]
    print(f"{col}: {len(outliers)} unusual values (outliers) found")

print("\n--- Data Spread Balance (Skewness) ---")
skewness = df[numerical_columns].skew().sort_values(ascending=False)
print(skewness)
```

## 6. Relationships Between Columns (Bivariate & Multivariate Analysis)

```python
plt.figure(figsize=(12, 8))
correlation = df.select_dtypes(include=['int64', 'float64']).corr()
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)
plt.title('How Number Columns Relate to Each Other')
plt.tight_layout()
plt.show()

print("\n--- How Strongly Each Feature Affects Price ---")
price_correlation = correlation['Price'].sort_values(ascending=False)
print(price_correlation)

for col in numerical_columns:
    if col != 'Price':
        plt.figure(figsize=(8, 5))
        sns.scatterplot(data=df, x=col, y='Price')
        plt.title(f'{col} vs Price')
        plt.tight_layout()
        plt.show()

for col in categorical_columns:
    plt.figure(figsize=(10, 5))
    sns.boxplot(data=df, x=col, y='Price')
    plt.title(f'{col} vs Price')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

sns.pairplot(df[numerical_columns], diag_kind='hist', corner=True)
plt.suptitle('Comparing All Number Columns Together', y=1.02)
plt.show()
```

## 7. Final Summary Report

```python
print("\n==============================================")
print("FINAL SUMMARY REPORT - PARASHURAM VADDEKUNTA")
print("==============================================")
print(f"Total Rows: {df.shape[0]} | Total Columns: {df.shape[1]}")
print(f"Number Columns: {len(numerical_columns)} | Text Columns: {len(categorical_columns)}")
print(f"Total Missing Values: {df.isnull().sum().sum()}")
print(f"Total Duplicate Rows: {df.duplicated().sum()}")
print("\n--- Car Price Details ---")
print(f"Average Price: {df['Price'].mean():.2f}")
print(f"Lowest Price: {df['Price'].min():.2f}")
print(f"Highest Price: {df['Price'].max():.2f}")
print("\n--- Top Features Related to Price ---")
print(price_correlation.drop('Price').abs().sort_values(ascending=False).head())
```
